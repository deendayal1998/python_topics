# Example 2: Create a Bank account
Task:
Defie a BankAccount class.
Add attributess for account holder name or balance.
Add methods to deposit withdraw, and check balance.

```python
class BankAccount:
    def __init__(self, holder_name, balance=0):
        self.holder_name = holder_name
        self.balance = balance

    def deposite(self, amount):
        self.balance += amount
        print(f"{amount} deposited, New balance: {self.balance}")

    def withdraw(self, amount):
        if amount<= self.balance:
            self.balance -= amount 
            print(f"{amount} withdraw, Remaining balance:{self.balance}")
        else:
            print("Insuficient funds!")

    def check_balance(self):
        return f"Account balance for {self.holder_name}: {self.balance}"

account = BankAccount("Deendayal", 1000)
account.deposite(500)
account.withdraw(300)
print(account.check_balance())
    
```

# Example 3: Create a Student Class
Task:
Define a Student class with attributes like name,age and marks
Add a method to display the student details
Add a method to check if the student passed(marks>=40).


```python
class Student:
    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks

    def display_details(self):
        print(f"Name :{self.name}, Age: {self.age}, Marks: {self.marks}")

    def has_passed(self):
        if self.marks>= 40:
            return "Passed"
        else:
            return "Failed"
# Test the Student class
student = Student("Riya", 22, 85) 
student.display_details() 
print(student.has_passed())
```

# Example4: Create a Library Sysytem
#Task:
#Define a Book class with attributes like , author, and availability.
#Add method to check availability and borrow/return the book.


```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.is_available = True

    def borrow(self):
        if self.is_available:
            self.is_available = False
            return f"You have borrowed '{self.title}'."
        else:
            return f"{self.title} is not available."

    def return_book(self):
        self.is_available = True
        return f"{self.title} has been returned, Thank you!"


book1 = Book("The Alchemist", "Paulo Coelho")
print(book1.borrow())
print(book1.borrow())
print(book1.return_book())
            
```

# Example 5: Create an Employee Management System


```python
# Task:
# Define an Employee class with attributes like name, ID, and salary.
# Add a method to calculate annual salary
# Add a method to display employee details
# Add two method to calculate salary on tha basis of hourly and day wise.
```

```python
class Employee:
    def __init__(self, name, ID, salary, time):
        self.name = name
        self.ID = ID
        self.salary = salary
        self.time = time

    def annual_salary(self):
        return self.salary*12

    def employee_details(self):
        return f"The employee Name :{self.name}, ID : {self.ID},  salary: {self.salary}"

    def freelance(self):
        return f"The salary of freelancer: {self.time*1000}"

    def monthly(self):
        return f"total salary for a month: {self.time*30*1000}"
        
employee = Employee("DD", 101, 25000, 8)
print(employee.annual_salary())
print(employee.employee_details())
print(employee.freelance())
print(employee.monthly())
```

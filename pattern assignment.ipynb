```python
num=1
for i in range(1,5):
    for j in range(1,i+1):
        print(num, end=" ")
        num+=1
    print("")
```

```python
for i in range(5,1,-1):
    num=4
    for j in range(i-1):
        print(num, end=" ")
        num-=1
    print("")
```

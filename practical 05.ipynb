```python
#String indexing and slicing

#numeric
#character-string,list,tuple
#'hello'
#'hello how are you'
```

```python
my_string="Hello"
print(my_string[0])
print(my_string[1])
print(my_string[4])
print(my_string[-2])
print(my_string[-1])
print(my_string[-4])
#indexing
```

```python
#slicing
my_string='Hello'
print(my_string[1:4])
print(my_string[:3])
print(my_string[2:])
print(my_string[0:])
```

```python
#step size
my_string='Hello'
print(my_string[0:4:2])
print(my_string[::-1])
```

```python
print(my_string[2:4:3])
print(my_string[0:3:2])
print(my_string[1:4:3])
print(my_string[0:2:2])
print(my_string[-1:-5:-2])
print(my_string[-1:-4:2])
```

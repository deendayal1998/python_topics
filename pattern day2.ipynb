```python
num=6
for i in range(1,num):
    x=4
    for j in range(i,num-1):
        print(" ", end=" ")
    for k in range(1,i+1):
        print(x,end=" ")
        x+=1
    print("")
```

```python
for i in range(1,5):
    for j in range(i,4):
        print(" ",end="")

    for k in range(1,i+1):
        print("*",end=" ")

    print("")
```

```python
for i in range(1,6):
    for j in range(1,i+1):
        if (j==1 or i==5 or i==j):
            print("*",end=" ")
        else:
            print(" ",end=" ")

    print()
```
